## NYC Taxi Trip Analysis
This notebook analyzes peak trip hours, average distances, and fares
using the Databricks sample NYC taxi dataset.

```python
# Databricks includes free sample datasets
df = spark.read.format("delta").load("/databricks-datasets/nyctaxi/tables/nyctaxi_yellow")
df.printSchema()
```

```python
display(df.limit(10))
```

```python
from pyspark.sql.functions import avg, count, hour

summary = df.groupBy(hour("pickup_datetime").alias("hour")) \
    .agg(
        count("*").alias("total_trips"),
        avg("trip_distance").alias("avg_distance_miles"),
        avg("total_amount").alias("avg_fare_usd")
    ) \
    .orderBy("hour")

display(summary)
```

```python
# The "Load" step — write transformed data to a permanent Delta table
summary.write.format("delta").mode("overwrite").saveAsTable("nyc_taxi_hourly_summary")
```
